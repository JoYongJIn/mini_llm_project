```python
text = """
나는 밥을 먹었다 나는 학교에 갔다 오늘은 날씨가 좋다
나는 커피를 마신다 그는 책을 읽는다 우리는 운동을 한다
"""
```

```python
import torch

# 단어 단위 토큰화
tokens = text.split() # 가장 단순한 토큰화 방법입니다. 공백을 기준으로 텍스트를 나누어 단어 단위로 토큰화합니다.
# split() 메서드는 문자열을 공백을 기준으로 나누어 리스트로 반환합니다. 
# 메서드와 함수의 차이는 메서드는 객체에 속한 함수입니다. 쉽게말해 .이 붙으면 메서드 입니다. 예를 들어, split()은 문자열 객체의 메서드입니다. 반면에 함수는 독립적으로 존재하는 코드 블록입니다. 예를 들어, print()는 내장 함수입니다.
# 내장함수와 함수의 차이는 내장 함수는 파이썬 인터프리터에 내장되어 있는 함수입니다. 예를 들어, print()는 내장 함수입니다. 반면에 함수는 사용자가 정의한 코드 블록입니다. 예를 들어, def my_function(): ...은 사용자가 정의한 함수입니다.
# vocab 생성
vocab = sorted(list(set(tokens)))
stoi = {w:i for i,w in enumerate(vocab)}
itos = {i:w for w,i in stoi.items()}

vocab_size = len(vocab)

def encode(s):
    return [stoi[w] for w in s.split()]

def decode(l):
    return " ".join([itos[i] for i in l])
```

```python
seq_len = 5 # 시퀀스 길이는 학습할 때 모델이 한 번에 볼 단어의 수입니다. 예를 들어, seq_len이 5이면 모델은 5개의 단어를 보고 다음 단어를 예측하도록 학습됩니다.

data = encode(text)

X, y = [], []

for i in range(len(data) - seq_len):
    X.append(data[i:i+seq_len])
    y.append(data[i+1:i+seq_len+1])

X = torch.tensor(X)
y = torch.tensor(y)
```

```python
def causal_mask(size):
    return torch.tril(torch.ones(size, size))
```

```python
import torch.nn as nn
import torch.nn.functional as F

class SelfAttention(nn.Module):
    def __init__(self, d_model):
        super().__init__()
        self.q = nn.Linear(d_model, d_model)
        self.k = nn.Linear(d_model, d_model)
        self.v = nn.Linear(d_model, d_model)

    def forward(self, x):
        B, T, D = x.shape

        Q = self.q(x)
        K = self.k(x)
        V = self.v(x)

        scores = Q @ K.transpose(-2, -1) / (D ** 0.5)

        mask = causal_mask(T).to(x.device)
        scores = scores.masked_fill(mask == 0, float('-inf'))

        attn = F.softmax(scores, dim=-1)

        out = attn @ V
        return out
```

```python
class Block(nn.Module):
    def __init__(self, d_model):
        super().__init__()
        self.attn = SelfAttention(d_model)
        self.ffn = nn.Sequential(
            nn.Linear(d_model, d_model*4),
            nn.ReLU(),
            nn.Linear(d_model*4, d_model)
        )
        self.ln1 = nn.LayerNorm(d_model)
        self.ln2 = nn.LayerNorm(d_model)

    def forward(self, x):
        x = self.ln1(x + self.attn(x))
        x = self.ln2(x + self.ffn(x))
        return x
```

```python
class GPT(nn.Module):
    def __init__(self, vocab_size, d_model, seq_len):
        super().__init__()
        self.token_emb = nn.Embedding(vocab_size, d_model) # 여기에 단어 임베딩 행렬이 생성됩니다. vocab_size는 단어 사전의 크기이고, d_model은 각 단어가 임베딩되는 벡터의 차원입니다.
        self.pos_emb = nn.Embedding(seq_len, d_model)

        self.blocks = nn.Sequential(
            Block(d_model),
            Block(d_model)
        )

        self.ln = nn.LayerNorm(d_model)
        self.head = nn.Linear(d_model, vocab_size)

    def forward(self, x):
        B, T = x.shape

        tok = self.token_emb(x)
        pos = self.pos_emb(torch.arange(T, device=x.device))

        x = tok + pos
        x = self.blocks(x)
        x = self.ln(x)

        return self.head(x)
```

```python
model = GPT(vocab_size, d_model=64, seq_len=seq_len)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
criterion = nn.CrossEntropyLoss()

for epoch in range(300):
    optimizer.zero_grad()

    out = model(X)

    loss = criterion(out.view(-1, vocab_size), y.view(-1))

    loss.backward()
    optimizer.step()

    if epoch % 50 == 0:
        print(epoch, loss.item())
```

```python
def generate(model, start, max_len=10):
    model.eval()

    tokens = encode(start)

    for _ in range(max_len):
        x = torch.tensor(tokens[-seq_len:]).unsqueeze(0)

        out = model(x)
        next_token = out[0, -1].argmax().item()

        tokens.append(next_token)

    return decode(tokens)
```

```python
# 단어 임베딩 행렬
print(model.token_emb.weight)
print(model.token_emb.weight.shape) # 16개 단어를 64차원으로 표현
```

```python
print(generate(model, "나는", 10))
```
