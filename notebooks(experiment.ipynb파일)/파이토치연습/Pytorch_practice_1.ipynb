```python
import torch
a = torch.tensor([1,2,3,4])
print(a)
print(type(a))
print(a.dtype)
print(a.shape)
b = torch.tensor([1,2,3.1,4])
print(b.dtype)
print(b)
```

```python
A = torch.tensor([[1,2,3],[3,4,5]])
print(A)
print(A.shape)
print(A.ndim) # number of dimension
print(A.numel()) #number of element
```

```python
print(torch.zeros(5))
print(torch.zeros_like(A))
print(torch.ones(5))
```

```python
print(torch.zeros(3,3))
print(torch.arange(3,10,2)) # range와 같은데 tensor로 만들어줌
print(torch.arange(0,1,0.1))
print(torch.linspace(0,1,10)) # linear space (0에서부터 1포함 10개로)
```

```python
a = torch.tensor([1,2,3])
b = torch.tensor([4,5,6])
c = a+b
print(c)
```

```python
A = torch.tensor([[1,2,3],[1,2,3]])
B = torch.tensor([[4,5,6],[1,1,1]])
C = A+B
D = A-B
print(C)
print(D)
```

```python
print(A*B) # Hadamard product라고 성분끼리의 곱으로 내적 연산은 @이다.
print(A/B)
print(B**2)
```

```python
A = torch.tensor([[1,2],[3,4]])
B = torch.tensor([[1,2],[3,4]])
print(A*B)
print(A@B) # 트랜스포머의 어텐션에서도 @연산을 사용한다.
```
