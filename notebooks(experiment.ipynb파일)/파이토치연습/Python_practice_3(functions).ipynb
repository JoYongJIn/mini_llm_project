```python
A = torch.randn(3,4)
print(A)
print(torch.max(A))
print(torch.max(A,dim=0)) # 열별로 최댓값을 구하고 그인덱스를 출력해준다.
print(torch.max(A,dim=1)) # 행별로 ~
print(torch.argmax(A))
print(torch.argmax(A,dim=0)) # 열별로 가장 큰값의 인덱스
print(torch.argmax(A,dim=1)) # 행별로 가장 큰값의 인덱스
```

```python
# 내적하는법

(1)
a = torch.tensor([1,2,3])
b = torch.tensor([2,2,1])
print(torch.sum(a*b))

(2)
a = a.reshape(3,1)
b = b.reshape(3,1)
print(a.transpose(1,0)@b) # 1번째 demension과 0번쨰 demension은 자리를 바꿔라
print(a.permute(1,0)@b) # permute는 premute(0번쨰에 몇번째 놈 오게할래,1번쨰에 몇번째 놈 오게할래)
print(a.T@b)
print(a.t()@b)
```

```python
# 잘하는 사람들은 간편하게 쓰기위해 축약된 표현도있다.
x = torch.randn(2,3,4,5,6)
#print(x)
#print(x[1,2,:,:,:]) #두 번째 배치의 세 번째 채널만 선택, 나머지는 그대로 유지
print(x[1,2,:,:,:].shape)
print(x[1,2,...].shape)
print(x[:,:,:,:,3].shape)
print(x[...,3].shape)
print(x[1,:,:,3,:].shape)
print(x[1,...,3,:].shape)
```

```python
# conacatenation 중요한 함수

A = torch.ones(3,4)
B = torch.zeros(3,4)
C = torch.vstack([A,B]) # vertical 세로방향으로 쌓는다
D = torch.hstack([A,B]) # horizontal 가로방향으로 쌓는다
E = torch.cat([A,B], dim=0) # 위보다 일반화된 표현으로 각각 dim=0 or dim=1에 따라 세로 가로방향으로 쌓는다.
F = torch.cat([A,B], dim=1)

print(C)
print(D)
print(E)
print(F)
```

```python
# squeeze & unsqueeze

A = torch.randn(1,1,1,3,1,1,1,4,1,1,1) # squeeze => 텐서에서 차원이 1인 축(dim) 을 자동으로 제거해주는 함수야 (축삭뻗음 효과등에 이용해보자)
print(A.shape)
print(A.squeeze().shape) # 1을 쫙 짜서 없앰
print(A.squeeze(dim=(0,2,4,5)).shape) # 지정한것만 없애줌
```

```python
A = torch.randn(3,4) # 반대로 unsqueeze는 가상의 한 차원을 내가 원한는 위치에 늘려준다.
print(A.unsqueeze(dim=0).shape)
print(A.unsqueeze(dim=1).shape)
print(A.unsqueeze(dim=2).shape)
print(A.reshape(1,3,4).shape) # 이렇게 해줘도 되지만 차원을 모두 써주어야하는것이 번거로움
print(A.reshape(3,1,4).shape)
print(A.reshape(3,4,1).shape) # 3행 4열자리 행렬을 깊이 축으로 쌓고싶을때
```

```python
A = torch.ones(3,4)
B = torch.zeros(3,4)
C = torch.stack([A,B]) # unsqueeze 말고 stack이용 가능 (차원을 추가해서 쌓아줌 때문에 A,B shape이 완전히 일치해야함,concat은 달라도 쌓을 차원 이외 차원만 맞아면됨
print(C)
```

```python
A = torch.tensor([[1,2],[3,4]])
B = A.clone() # B = A 라고만 하면 아예 둘이 같아져 주소까지 같이가져와 하나를 수정하면 하나도 수정이된다 이것을 막기위한것이 clone이다.
B[0,0] = 100

print(B)
print(A)
```
