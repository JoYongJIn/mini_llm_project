```python
import random
import math

# ----- 확률 파라미터 -----
p = 1 / 125
q = 1 - p

total = int(input("얼마를 충전하시겠습니까? "))
jackpot_num_list = []
previous_betting = 0

# 구간 정의(양끝 포함)
intervals = [(1, 28), (29, 64), (65, 115), (116, 200), (201, float('inf'))]

def make_jackpot_num():
    """기하분포(p=1/125)를 시뮬레이션해서 첫 잭팟 시점을 반환"""
    jackpot_num = 1
    while True:
        if random.randint(1, 125) == 1:
            return jackpot_num
        jackpot_num += 1

def classify_interval(x, intervals):
    """값 x가 속한 구간의 인덱스(0~len-1)를 반환"""
    for i, (lo, hi) in enumerate(intervals):
        if lo <= x <= hi:
            return i
    return None

def signed_streaks_by_interval(nums, intervals):
    """각 구간의 '부호 있는 스트릭' 계산"""
    seq = [classify_interval(x, intervals) for x in nums]
    m = len(intervals)
    L = len(seq)
    if L == 0:
        return [0] * m

    streaks = []
    for target in range(m):
        if seq[-1] == target:
            r = 0
            j = L - 1
            while j >= 0 and seq[j] == target:
                r += 1
                j -= 1
            streaks.append(+r)
        else:
            g = 0
            j = L - 1
            while j >= 0 and seq[j] != target:
                g += 1
                j -= 1
            streaks.append(-g)
    return streaks

def count_all_intervals(nums, intervals):
    """전체 잭팟 기록에서 각 구간의 누적 출현 횟수 계산"""
    counts = [0] * len(intervals)
    for x in nums:
        idx = classify_interval(x, intervals)
        if idx is not None:
            counts[idx] += 1
    return counts, len(nums)

def get_disappeared_bin(nums, intervals, window_size=5):
    """최근 window_size개 구간 중 사라진 구간 계산"""
    if len(nums) < window_size:
        return None
    old_value = nums[-window_size]
    disappeared_bin = classify_interval(old_value, intervals)
    return disappeared_bin

def moving_average(data, n):
    """최근 n개 평균"""
    if len(data) == 0:
        return 0
    subset = data[-n:] if len(data) >= n else data
    return round(sum(subset) / len(subset))

def interval_distribution(nums, intervals, n):
    """최근 n개 구간 분포"""
    recent = nums[-n:]
    counts = [0] * len(intervals)
    for num in recent:
        idx = classify_interval(num, intervals)
        if idx is not None:
            counts[idx] += 1
    return counts

def slot_machine():
    global total, previous_betting, jackpot_num_list
    jackpot_num = make_jackpot_num()
    play_num = 0

    while True:
        betting_str = input("얼마를 배팅 하시겠습니까? ")

        if betting_str == '':
            if previous_betting == 0:
                print("이전 배팅 금액이 없습니다. 배팅 금액을 입력해주세요.")
                continue
            else:
                betting = previous_betting
        else:
            try:
                betting = int(betting_str)
                if betting <= 0:
                    print("배팅 금액은 0보다 커야 합니다.")
                    continue
            except ValueError:
                print("유효한 배팅 금액을 입력해주세요.")
                continue

        if betting > total:
            print("배팅 금액이 잔액을 초과합니다.")
            continue

        previous_betting = betting
        play_num += 1
        total -= betting

        probability = (1 - (q ** play_num)) * 100

        # 💡 구간분포: 5, 10, 15, 20개 기준으로 계산
        interval_counts = {n: interval_distribution(jackpot_num_list, intervals, n) for n in [5, 10, 15, 20]}

        # 💡 이동평균: 5, 10, 15, 20개 기준으로 계산
        averages = {n: moving_average(jackpot_num_list, n) for n in [5, 10, 15, 20]}

        signed_streaks = signed_streaks_by_interval(jackpot_num_list, intervals)
        total_counts, n_total = count_all_intervals(jackpot_num_list, intervals)

        disappeared_bin = get_disappeared_bin(jackpot_num_list, intervals, window_size=5)

        print("\n")
        print(f"현재 {play_num}회 배팅 중입니다. total: {total}")
        for n in [5, 10, 15, 20]:
            print(f"구간분포(최근{n}) {interval_counts[n]}")
        print(f"당첨확률(기하CDF): {probability:.2f}%")

        avg_str = ", ".join([f"{n}개={averages[n]}" for n in [5, 10, 15, 20]])
        print(f"최근 평균: {avg_str}")
        print(f"구간별 연속정도 : {signed_streaks}")
        print(f"전체 {n_total}개 구간별 누적 횟수: {total_counts}")

        if disappeared_bin is not None:
            lo, hi = intervals[disappeared_bin]
            print(f"최근 5회 기준 사라진 구간: {disappeared_bin + 1} ({lo}~{hi})")
        else:
            print(f"최근 5회 기준 사라진 구간: 없음 (데이터 부족)")

        print(f"(1,28)(29,64)(65,115)(116,200)(201), 이전 {jackpot_num_list[-5:]}")

        # 🎰 잭팟 처리
        if play_num == jackpot_num:
            total += betting * 110
            jackpot_num_list.append(play_num)
            print(f"이야~ 대박!🧞‍♂️ 축하합니다! 잭팟입니다! total: {total}")
            print("🧚‍♂️ 🧚‍♀️")
            break

while True:
    slot_machine()
```

```python
import random
import math

# ----- 확률 파라미터 -----
p = 1 / 125
q = 1 - p

total = int(input("얼마를 충전하시겠습니까? "))
jackpot_num_list = []
previous_betting = 0

# 구간 정의(양끝 포함)
intervals = [(1, 28), (29, 64), (65, 115), (116, 200), (201, float('inf'))]


# ------------------------------
# 임베딩 벡터 관련 함수들
# ------------------------------

def make_jackpot_num():
    """기하분포(p=1/125)를 시뮬레이션해서 첫 잭팟 시점을 반환"""
    jackpot_num = 1
    while True:
        if random.randint(1, 125) == 1:
            return jackpot_num
        jackpot_num += 1

def classify_interval(x, intervals):
    """값 x가 속한 구간의 인덱스(0~len-1)를 반환"""
    for i, (lo, hi) in enumerate(intervals):
        if lo <= x <= hi:
            return i
    return None

def signed_streaks_by_interval(nums, intervals):
    """각 구간의 '부호 있는 스트릭' 계산"""
    seq = [classify_interval(x, intervals) for x in nums]
    m = len(intervals)
    L = len(seq)
    if L == 0:
        return [0] * m

    streaks = []
    for target in range(m):
        if seq[-1] == target:
            r = 0
            j = L - 1
            while j >= 0 and seq[j] == target:
                r += 1
                j -= 1
            streaks.append(+r)
        else:
            g = 0
            j = L - 1
            while j >= 0 and seq[j] != target:
                g += 1
                j -= 1
            streaks.append(-g)
    return streaks

def count_all_intervals(nums, intervals):
    """전체 잭팟 기록에서 각 구간의 누적 출현 횟수 계산"""
    counts = [0] * len(intervals)
    for x in nums:
        idx = classify_interval(x, intervals)
        if idx is not None:
            counts[idx] += 1
    return counts, len(nums)

def get_disappeared_bin(nums, intervals, window_size=5):
    """최근 window_size개 구간 중 사라진 구간 계산"""
    if len(nums) < window_size:
        return None
    old_value = nums[-window_size]
    disappeared_bin = classify_interval(old_value, intervals)
    return disappeared_bin

def moving_average(data, n):
    """최근 n개 평균"""
    if len(data) == 0:
        return 0
    subset = data[-n:] if len(data) >= n else data
    return round(sum(subset) / len(subset))

def interval_distribution(nums, intervals, n):
    """최근 n개 구간 분포"""
    recent = nums[-n:]
    counts = [0] * len(intervals)
    for num in recent:
        idx = classify_interval(num, intervals)
        if idx is not None:
            counts[idx] += 1
    return counts

# ------------------------------
# 30차원 임베딩 벡터 생성 함수
# ------------------------------

def create_embedding_vector(jackpot_num_list, intervals, play_num):
    """
    의미 기반 30차원 임베딩 벡터 생성 (모든 값 소수 둘째 자리 반올림)
    """

    # 1. 최근 구간 분포(5,10,15,20)
    interval_features = []
    for n in [5, 10, 15, 20]:
        interval_features.extend(interval_distribution(jackpot_num_list, intervals, n))

    # 2. 최근 평균 (5, 10, 15, 20)
    avg_features = [moving_average(jackpot_num_list, n) for n in [5, 10, 15, 20]]

    # 3. 구간별 연속정도 (signed streak)
    streak_features = signed_streaks_by_interval(jackpot_num_list, intervals)

    # 4. 당첨 확률 (기하 CDF)
    p = 1/125
    q = 1 - p
    probability_cdf = (1 - (q ** play_num))

    # --- 하나의 임베딩 벡터로 합치기 ---
    embedding_vector = []
    embedding_vector.extend(interval_features)
    embedding_vector.extend(avg_features)
    embedding_vector.extend(streak_features)
    embedding_vector.append(probability_cdf)

    # 🔥 모든 요소 소수 둘째 자리 반올림
    embedding_vector = [round(float(x), 2) for x in embedding_vector]

    return embedding_vector


# ------------------------------
# 메인 게임 함수
# ------------------------------

def slot_machine():
    global total, previous_betting, jackpot_num_list
    jackpot_num = make_jackpot_num()
    play_num = 0

    while True:
        betting_str = input("얼마를 배팅 하시겠습니까? ")

        if betting_str == '':
            if previous_betting == 0:
                print("이전 배팅 금액이 없습니다. 배팅 금액을 입력해주세요.")
                continue
            else:
                betting = previous_betting
        else:
            try:
                betting = int(betting_str)
                if betting <= 0:
                    print("배팅 금액은 0보다 커야 합니다.")
                    continue
            except ValueError:
                print("유효한 배팅 금액을 입력해주세요.")
                continue

        if betting > total:
            print("배팅 금액이 잔액을 초과합니다.")
            continue

        previous_betting = betting
        play_num += 1
        total -= betting

        probability = (1 - (q ** play_num)) * 100

        # 💡 구간분포: 5, 10, 15, 20개 기준으로 계산
        interval_counts = {n: interval_distribution(jackpot_num_list, intervals, n) for n in [5, 10, 15, 20]}

        # 💡 이동평균: 5, 10, 15, 20개 기준으로 계산
        averages = {n: moving_average(jackpot_num_list, n) for n in [5, 10, 15, 20]}

        signed_streaks = signed_streaks_by_interval(jackpot_num_list, intervals)
        total_counts, n_total = count_all_intervals(jackpot_num_list, intervals)

        disappeared_bin = get_disappeared_bin(jackpot_num_list, intervals, window_size=5)

        print("\n")
        print(f"현재 {play_num}회 배팅 중입니다. total: {total}")
        for n in [5, 10, 15, 20]:
            print(f"구간분포(최근{n}) {interval_counts[n]}")
        print(f"당첨확률(기하CDF): {probability:.2f}%")

        avg_str = ", ".join([f"{n}개={averages[n]}" for n in [5, 10, 15, 20]])
        print(f"최근 평균: {avg_str}")
        print(f"구간별 연속정도 : {signed_streaks}")
        print(f"전체 {n_total}개 구간별 누적 횟수: {total_counts}")

        if disappeared_bin is not None:
            lo, hi = intervals[disappeared_bin]
            print(f"최근 5회 기준 사라진 구간: {disappeared_bin + 1} ({lo}~{hi})")
        else:
            print(f"최근 5회 기준 사라진 구간: 없음 (데이터 부족)")

        print(f"(1,28)(29,64)(65,115)(116,200)(201), 이전 {jackpot_num_list[-5:]}")

        # -------------------------
        # 📌 임베딩 벡터 출력 추가됨!
        # -------------------------
        emb = create_embedding_vector(jackpot_num_list, intervals, play_num)
        print("\n📌 생성된 임베딩벡터 (길이:", len(emb), ")")
        print(emb)
        # -------------------------

        # 🎰 잭팟 처리
        if play_num == jackpot_num:
            total += betting * 110
            jackpot_num_list.append(play_num)
            print(f"이야~ 대박!🧞‍♂️ 축하합니다! 잭팟입니다! total: {total}")
            print("🧚‍♂️ 🧚‍♀️")
            break


# ------------------------------
# 게임 반복 실행
# ------------------------------

while True:
    slot_machine()
```

# 1. 구간분포 확률로 매핑

```python
# 최근 5개 구간분포 확률화

import random
from collections import Counter
import pandas as pd

# ----- 확률 파라미터 -----
p = 1 / 125
q = 1 - p

# ✅ 구간 정의 (마지막 구간을 무한대로 확장)
intervals = [
    (1, 28),
    (29, 64),
    (65, 115),
    (116, 200),
    (201, float('inf'))  # ← 수정된 부분
]

# ----- 잭팟 발생 시점 (기하분포 기반) -----
def make_jackpot_num():
    jackpot_num = 1
    while True:
        if random.randint(1, 125) == 1:
            return jackpot_num
        jackpot_num += 1

# ----- 어떤 값이 어느 구간에 속하는지 -----
def classify_interval(x, intervals):
    for i, (lo, hi) in enumerate(intervals):
        if lo <= x <= hi:
            return i
    return None

# ----- 최근 n개의 잭팟으로 구간분포(히스토그램) 계산 -----
def interval_histogram(nums, intervals, n=5):
    recent = nums[-n:]
    counts = [0] * len(intervals)
    for x in recent:
        idx = classify_interval(x, intervals)
        if idx is not None:
            counts[idx] += 1
    return tuple(counts)

# ----- 시뮬레이션 -----
def simulate_pattern_counts(n_jackpots=10000, window_size=5):
    jackpot_nums = []
    pattern_counter = Counter()

    for _ in range(n_jackpots):
        jackpot = make_jackpot_num()
        jackpot_nums.append(jackpot)

        if len(jackpot_nums) >= window_size:
            pattern = interval_histogram(jackpot_nums, intervals, n=window_size)
            if sum(pattern) == window_size:  # 이 부분 그대로 유지
                pattern_counter[pattern] += 1

    return pattern_counter

# ----- 실행 -----
n = int(input("생성할 잭팟 개수를 입력하세요: "))
pattern_counts = simulate_pattern_counts(n)

# ----- DataFrame으로 정리 (등장횟수 내림차순) -----
total = sum(pattern_counts.values())

df = pd.DataFrame(
    [(pattern, count, count / total * 100) for pattern, count in pattern_counts.items()],
    columns=["Pattern", "Count", "Probability(%)"]
).sort_values("Probability(%)", ascending=False).reset_index(drop=True)

# ----- 전체 출력 -----
pd.set_option("display.max_rows", None)
print("\n=== 전체 구간분포 패턴 등장 확률 (%) ===")
print(df)

print(f"\n✅ 총 {len(df)}가지 패턴이 등장했습니다. (전체 {total}회)")

# 상위 30개정도가 50%를 차지한다.
```

```python
# 최근 10개 구간분포 확률화

import random
from collections import Counter
import pandas as pd

# ----- 확률 파라미터 -----
p = 1 / 125
q = 1 - p

# ✅ 구간 정의 (마지막 구간을 무한대로 확장)
intervals = [
    (1, 28),
    (29, 64),
    (65, 115),
    (116, 200),
    (201, float('inf'))
]

# ----- 잭팟 발생 시점 (기하분포 기반) -----
def make_jackpot_num():
    jackpot_num = 1
    while True:
        if random.randint(1, 125) == 1:
            return jackpot_num
        jackpot_num += 1

# ----- 어떤 값이 어느 구간에 속하는지 -----
def classify_interval(x, intervals):
    for i, (lo, hi) in enumerate(intervals):
        if lo <= x <= hi:
            return i
    return None

# ----- 최근 n개의 잭팟으로 구간분포(히스토그램) 계산 -----
def interval_histogram(nums, intervals, n=10):   # ← 🔥 여기 10으로 변경
    recent = nums[-n:]
    counts = [0] * len(intervals)
    for x in recent:
        idx = classify_interval(x, intervals)
        if idx is not None:
            counts[idx] += 1
    return tuple(counts)

# ----- 시뮬레이션 -----
def simulate_pattern_counts(n_jackpots=10000, window_size=10):  # ← 🔥 여기 10으로 변경
    jackpot_nums = []
    pattern_counter = Counter()

    for _ in range(n_jackpots):
        jackpot = make_jackpot_num()
        jackpot_nums.append(jackpot)

        if len(jackpot_nums) >= window_size:
            pattern = interval_histogram(jackpot_nums, intervals, n=window_size)
            if sum(pattern) == window_size:  
                pattern_counter[pattern] += 1

    return pattern_counter

# ----- 실행 -----
n = int(input("생성할 잭팟 개수를 입력하세요: "))
pattern_counts = simulate_pattern_counts(n)

# ----- DataFrame으로 정리 (등장횟수 내림차순) -----
total = sum(pattern_counts.values())

df = pd.DataFrame(
    [(pattern, count, count / total * 100) for pattern, count in pattern_counts.items()],
    columns=["Pattern", "Count", "Probability(%)"]
).sort_values("Probability(%)", ascending=False).reset_index(drop=True)

# ----- 전체 출력 -----
pd.set_option("display.max_rows", None)
print("\n=== 전체 구간분포 패턴 등장 확률 (최근 10개 기준) (%) ===")
print(df)

print(f"\n✅ 총 {len(df)}가지 패턴이 등장했습니다. (전체 {total}회)")
```

```python
# 모든 구간분포 한번에 확률화

import random
from collections import Counter
import pandas as pd

# ----- 확률 파라미터 -----
p = 1 / 125
q = 1 - p

# ----- 구간 정의 -----
intervals = [
    (1, 28),
    (29, 64),
    (65, 115),
    (116, 200),
    (201, float('inf'))
]

# ----- 잭팟 생성 -----
def make_jackpot_num():
    jackpot_num = 1
    while True:
        if random.randint(1, 125) == 1:
            return jackpot_num
        jackpot_num += 1

# ----- 구간 판별 -----
def classify_interval(x, intervals):
    for i, (lo, hi) in enumerate(intervals):
        if lo <= x <= hi:
            return i
    return None

# ----- 최근 n개 구간분포 패턴 -----
def interval_histogram(nums, intervals, n):
    recent = nums[-n:]
    counts = [0] * len(intervals)
    for x in recent:
        idx = classify_interval(x, intervals)
        if idx is not None:
            counts[idx] += 1
    return tuple(counts)

# ----- 패턴 시뮬레이션 -----
def simulate_pattern_counts(n_jackpots, window_size):
    jackpot_nums = []
    pattern_counter = Counter()

    for _ in range(n_jackpots):
        jackpot = make_jackpot_num()
        jackpot_nums.append(jackpot)

        if len(jackpot_nums) >= window_size:
            pattern = interval_histogram(jackpot_nums, intervals, window_size)
            if sum(pattern) == window_size:
                pattern_counter[pattern] += 1

    return pattern_counter

# ----- 확률 테이블로 변환 -----
def make_prob_table(pattern_counter):
    total = sum(pattern_counter.values())
    table = {pattern: round(count / total, 8) for pattern, count in pattern_counter.items()}
    return table

# ----- DF 출력 -----
def show_table(prob_table, label):
    df = pd.DataFrame([(pattern, prob) for pattern, prob in prob_table.items()],
                      columns=["Pattern", "Probability"]) \
         .sort_values("Probability", ascending=False) \
         .reset_index(drop=True)

    print(f"\n=== 🔥 {label} 구간분포 확률 테이블 ===")
    print(df)
    print(f"총 {len(df)}가지 패턴\n")


# ============================================================
# 🔥 통합 실행
# ============================================================

N = int(input("생성할 잭팟 개수를 입력하세요: "))
windows = [5, 10, 15, 20]

prob_tables = {}

for w in windows:
    patterns = simulate_pattern_counts(N, window_size=w)
    prob_tables[w] = make_prob_table(patterns)
    show_table(prob_tables[w], f"최근 {w}개")

print("\n🎯 모든 확률 테이블 생성 완료!")
```

### 구간분포 5,10,15,20모두 확률화
- 같은 패턴끼리 확률 평균내 매핑시키면 더 확실해짐

```python
# 모든경우 구간분포 백분율 확률화

import random
from collections import Counter
import pandas as pd

# ----- 확률 파라미터 -----
p = 1 / 125
q = 1 - p

# ----- 구간 정의 -----
intervals = [
    (1, 28),
    (29, 64),
    (65, 115),
    (116, 200),
    (201, float('inf'))
]

# ----- 잭팟 생성 -----
def make_jackpot_num():
    jackpot_num = 1
    while True:
        if random.randint(1, 125) == 1:
            return jackpot_num
        jackpot_num += 1

# ----- 구간 판별 -----
def classify_interval(x, intervals):
    for i, (lo, hi) in enumerate(intervals):
        if lo <= x <= hi:
            return i
    return None

# ----- 최근 n개 구간분포 패턴 -----
def interval_histogram(nums, intervals, n):
    recent = nums[-n:]
    counts = [0] * len(intervals)
    for x in recent:
        idx = classify_interval(x, intervals)
        if idx is not None:
            counts[idx] += 1
    return tuple(counts)

# ----- 시뮬레이션 -----
def simulate_pattern_counts(n_jackpots, window_size):
    jackpot_nums = []
    pattern_counter = Counter()

    for _ in range(n_jackpots):
        jackpot = make_jackpot_num()
        jackpot_nums.append(jackpot)

        if len(jackpot_nums) >= window_size:
            pattern = interval_histogram(jackpot_nums, intervals, window_size)
            if sum(pattern) == window_size:
                pattern_counter[pattern] += 1

    return pattern_counter

# ----- 확률 테이블로 변환 (퍼센트 단위) -----
def make_percent_table(pattern_counter):
    total = sum(pattern_counter.values())
    table = {}

    for pattern, count in pattern_counter.items():
        percent = (count / total) * 100
        table[pattern] = round(percent, 4)  # 퍼센트 단위
    return table

# ----- 출력 -----
def show_table(percent_table, label):
    df = pd.DataFrame([(pattern, percent_table[pattern]) for pattern in percent_table],
                      columns=["Pattern", "Probability(%)"]) \
         .sort_values("Probability(%)", ascending=False) \
         .reset_index(drop=True)

    print(f"\n=== 🔥 {label} 구간분포 확률 테이블 ===")
    print(df)
    print(f"총 {len(df)}가지 패턴\n")

# ============================================================
# 🔥 통합 실행
# ============================================================

N = int(input("생성할 잭팟 개수를 입력하세요: "))
windows = [5, 10, 15, 20]

percent_tables = {}

for w in windows:
    patterns = simulate_pattern_counts(N, window_size=w)
    percent_tables[w] = make_percent_table(patterns)
    show_table(percent_tables[w], f"최근 {w}개")

print("\n🎯 모든 확률 테이블 생성 완료!")
```

# 2. 평균으로부터 떨어진 정도 확률로 매핑

```python
import random
import math
from collections import Counter
import pandas as pd

# ===== 설정 =====
p = 1 / 125
EXPECTED_MEAN = 125

# 🔹 구간을 양수/음수로 나눈다
DIFF_BINS = [
    (-float('inf'), -201),
    (-200, -101),
    (-100, -51),
    (-50, -26),
    (-25, -1),
    (0, 0),
    (1, 25),
    (26, 50),
    (51, 100),
    (101, 200),
    (201, float('inf'))
]

# ===== 잭팟 생성 =====
def make_jackpot_num():
    jackpot_num = 1
    while True:
        if random.randint(1, 125) == 1:
            return jackpot_num
        jackpot_num += 1

# ===== 이동 평균 =====
def moving_average(nums, window):
    """최근 window개 평균을 정수로 반올림"""
    if len(nums) < window:
        return None
    return round(sum(nums[-window:]) / window)

# ===== 구간 분류 =====
def classify_signed_difference(diff):
    for i, (lo, hi) in enumerate(DIFF_BINS):
        if lo <= diff <= hi:
            return i
    return None

# ===== 시뮬레이션 =====
def simulate_signed_diff_bins(n_jackpots):
    jackpot_nums = []

    # 5, 10, 15, 20 평균을 위한 카운터 추가
    bins_5, bins_10, bins_15, bins_20 = Counter(), Counter(), Counter(), Counter()
    anomalies = {"5": [], "10": [], "15": [], "20": []}

    for round_idx in range(1, n_jackpots + 1):
        jackpot = make_jackpot_num()
        jackpot_nums.append(jackpot)

        for window, bins, key in [
            (5,  bins_5,  "5"),
            (10, bins_10, "10"),
            (15, bins_15, "15"),
            (20, bins_20, "20")
        ]:
            avg = moving_average(jackpot_nums, window)
            if avg is None:
                anomalies[key].append((round_idx, avg, "None(데이터부족)"))
                continue

            diff = round(avg - EXPECTED_MEAN)

            if math.isnan(diff) or math.isinf(diff):
                anomalies[key].append((round_idx, diff, "NaN/Inf"))
                continue

            bin_idx = classify_signed_difference(diff)
            if bin_idx is not None:
                bins[bin_idx] += 1
            else:
                anomalies[key].append((round_idx, diff, "구간없음"))

    return bins_5, bins_10, bins_15, bins_20, anomalies


# ===== 결과 요약 =====
def summarize(counter, label, anomaly_list):
    total_valid = sum(counter.values())
    total = total_valid + len(anomaly_list)
    data = []

    for i, (lo, hi) in enumerate(DIFF_BINS):
        range_label = f"{int(lo) if lo != -float('inf') else '-∞'}~{int(hi) if hi != float('inf') else '+∞'}"
        data.append([
            range_label,
            counter[i],
            round((counter[i] / total_valid * 100), 3) if total_valid > 0 else 0
        ])

    df = pd.DataFrame(data, columns=["Difference Range", "Count", "Probability(%)"])

    print(f"\n=== {label} ===")
    print(f"전체 시행 수: {total}")
    print(f"정상 데이터: {total_valid}, 이상 데이터: {len(anomaly_list)} ({round(len(anomaly_list)/total*100,3)}%)")
    print(df)

    if len(anomaly_list) > 0:
        print("\n[이상 데이터 예시 5개]")
        for row in anomaly_list[:5]:
            print(f"회차 {row[0]:5d} | avg={row[1]} | 유형={row[2]}")
    else:
        print("\n[이상 데이터 없음 ✅]")


# ===== 실행 =====
N = int(input("생성할 잭팟 개수를 입력하세요: "))
bins_5, bins_10, bins_15, bins_20, anomalies = simulate_signed_diff_bins(N)

summarize(bins_5,  "임베딩 16 (최근5 평균)", anomalies["5"])
summarize(bins_10, "임베딩 17 (최근10 평균)", anomalies["10"])
summarize(bins_15, "임베딩 18 (최근15 평균)", anomalies["15"])
summarize(bins_20, "임베딩 19 (최근20 평균)", anomalies["20"])

print("\n✅ 평균이 125보다 작으면 음수, 크면 양수로 차이를 구분했습니다.")
```

### 해당부분을 확률로 매핑시켜주는 딕셔너리 코드

```python
# ======================================================
# 🔥 평균 차이(diff-bin) 확률 테이블 자동 생성 코드 (5,10,15,20)
# ======================================================

def make_diff_prob_table(counter):
    """Counter(bin_index → count)를 확률 딕셔너리로 변환"""
    total = sum(counter.values())
    diff_prob_table = {}

    for bin_index, count in counter.items():
        prob = count / total if total > 0 else 0
        diff_prob_table[bin_index] = round(prob, 8)

    return diff_prob_table


# ★★★★★ 새로 추가된 부분: bins_15 포함 ★★★★★
diff_prob_5  = make_diff_prob_table(bins_5)
diff_prob_10 = make_diff_prob_table(bins_10)
diff_prob_15 = make_diff_prob_table(bins_15)
diff_prob_20 = make_diff_prob_table(bins_20)


# 보기 좋게 출력
print("\n=== 🔥 평균 diff-bin 확률 테이블 (최근5) ===")
for k in sorted(diff_prob_5.keys()):
    print(f"bin {k:2d}: {diff_prob_5[k]}")

print("\n=== 🔥 평균 diff-bin 확률 테이블 (최근10) ===")
for k in sorted(diff_prob_10.keys()):
    print(f"bin {k:2d}: {diff_prob_10[k]}")

print("\n=== 🔥 평균 diff-bin 확률 테이블 (최근15) ===")
for k in sorted(diff_prob_15.keys()):
    print(f"bin {k:2d}: {diff_prob_15[k]}")

print("\n=== 🔥 평균 diff-bin 확률 테이블 (최근20) ===")
for k in sorted(diff_prob_20.keys()):
    print(f"bin {k:2d}: {diff_prob_20[k]}")
```

```python
import random
import matplotlib.pyplot as plt
import numpy as np

# ----- Probability parameter -----
p = 1 / 125
q = 1 - p

def make_jackpot_num():
    jackpot_num = 1
    while True:
        if random.randint(1, 125) == 1:
            return jackpot_num
        jackpot_num += 1

def moving_average_last_n(data, n=5):
    if len(data) < n:
        return None
    return sum(data[-n:]) / n

def compute_pdf(values, bins=30):
    counts, bin_edges = np.histogram(values, bins=bins)
    total = np.sum(counts)
    pdf = counts / total
    return pdf, bin_edges, counts

def get_probability(avg_value, pdf, bin_edges):
    for i in range(len(bin_edges)-1):
        if bin_edges[i] <= avg_value < bin_edges[i+1]:
            lo = int(round(bin_edges[i]))
            hi = int(round(bin_edges[i+1]))
            return pdf[i], (lo, hi), i
    lo = int(round(bin_edges[-2]))
    hi = int(round(bin_edges[-1]))
    return pdf[-1], (lo, hi), len(bin_edges)-2

def simulate_and_plot(num_simulations, window_size=5, bins=30):
    jackpot_nums = []
    avg_history = []

    for _ in range(num_simulations):
        jn = make_jackpot_num()
        jackpot_nums.append(jn)

        avg_value = moving_average_last_n(jackpot_nums, window_size)
        if avg_value is not None:
            avg_history.append(avg_value)

    plt.figure(figsize=(10, 5))
    plt.hist(avg_history, bins=bins, edgecolor='black', alpha=0.75)
    plt.title(f"Distribution of Recent {window_size}-Number Averages ({num_simulations} sims)")
    plt.xlabel(f"Average value (recent {window_size})")
    plt.ylabel("Frequency")
    plt.grid(alpha=0.3)
    plt.show()

    pdf, bin_edges, counts = compute_pdf(avg_history, bins=bins)

    print("\n=== 확률 구간표 (자연수 반올림된 구간) ===")
    for i in range(bins):
        lo = int(round(bin_edges[i]))
        hi = int(round(bin_edges[i+1]))
        print(f"Bin {i:2d}:  [{lo:3d} ~ {hi:3d}]   Count = {counts[i]:6d}   Prob = {pdf[i]*100:6.3f}%")

    return avg_history, pdf, bin_edges

# Run
num_sim = int(input("Enter number of simulations: "))
avg_history, pdf, bins = simulate_and_plot(num_sim, window_size=5, bins=30)

test_avg = 125
prob, (lo, hi), idx = get_probability(test_avg, pdf, bins)

print(f"\n입력 평균값 {test_avg} → Bin {idx}")
print(f"구간 = [{lo} ~ {hi}]")
print(f"확률 = {prob*100:.3f}%")
```

# 3. 구간별연속정도의 숫자들이 의미하는 빈도수 확률적해석

```python
import random
from collections import Counter
import pandas as pd

# ----- 확률 파라미터 -----
p = 1 / 125
q = 1 - p

# ----- 구간 정의 -----
intervals = [
    (1, 28),
    (29, 64),
    (65, 115),
    (116, 200),
    (201, float('inf'))
]

# ----- 잭팟 시점 생성 -----
def make_jackpot_num():
    jackpot_num = 1
    while True:
        if random.randint(1, 125) == 1:
            return jackpot_num
        jackpot_num += 1

# ----- 구간 판별 -----
def classify_interval(x, intervals):
    for i, (lo, hi) in enumerate(intervals):
        if lo <= x <= hi:
            return i
    return None

# ----- 부호 있는 연속도 계산 -----
def signed_run_length_by_interval(nums, intervals):
    """
    기존 streaks → '연속도'
    +값 = 최근 해당 구간이 연속으로 나온 횟수
    -값 = 최근 해당 구간이 얼마나 오래 안나오고 있는지(결핍 정도)
    """
    seq = [classify_interval(x, intervals) for x in nums]
    m = len(intervals)
    L = len(seq)
    if L == 0:
        return [0] * m

    run_values = []
    for target in range(m):
        if seq[-1] == target:
            # 최근 연속 출현 (+)
            r = 0
            j = L - 1
            while j >= 0 and seq[j] == target:
                r += 1
                j -= 1
            run_values.append(+r)
        else:
            # 최근에 안 나옴 (−)
            g = 0
            j = L - 1
            while j >= 0 and seq[j] != target:
                g += 1
                j -= 1
            run_values.append(-g)
    return run_values

# ----- 시뮬레이션 -----
def simulate_run_value_counts(n_jackpots=10000):
    jackpot_nums = []
    value_counter = Counter()

    for _ in range(n_jackpots):
        jackpot = make_jackpot_num()
        jackpot_nums.append(jackpot)

        if len(jackpot_nums) >= 5:
            run_values = signed_run_length_by_interval(jackpot_nums, intervals)
            for val in run_values:
                value_counter[val] += 1  # 연속도 값별 등장 횟수 누적

    return value_counter

# ----- 실행 -----
n = int(input("생성할 잭팟 개수를 입력하세요: "))
value_counts = simulate_run_value_counts(n)

# ----- DataFrame 변환 -----
total = sum(value_counts.values())  # 전체 연속도 값 개수 합계
df = pd.DataFrame(
    [(val, count, count / total * 100) for val, count in value_counts.items()],
    columns=["Run Value(연속도)", "Count", "Probability(%)"]
).sort_values("Probability(%)", ascending=False).reset_index(drop=True)

# ----- 전체 출력 -----
pd.set_option("display.max_rows", None)
print("\n=== 연속도 값별 등장 확률 (내림차순) ===")
print(df)
print(f"\n✅ 총 {len(df)}가지 연속도 값이 등장했습니다. (전체 {total}회)")
```

```python
import random
from collections import Counter
import pandas as pd

# ----- 확률 파라미터 -----
p = 1 / 125
q = 1 - p

# ----- 구간 정의 -----
intervals = [
    (1, 28),
    (29, 64),
    (65, 115),
    (116, 200),
    (201, float('inf'))
]

# ----- 잭팟 시점 생성 -----
def make_jackpot_num():
    jackpot_num = 1
    while True:
        if random.randint(1, 125) == 1:
            return jackpot_num
        jackpot_num += 1

# ----- 구간 분류 -----
def classify_interval(x, intervals):
    for i, (lo, hi) in enumerate(intervals):
        if lo <= x <= hi:
            return i
    return None

# ----- 연속도 계산 -----
def signed_run_length_by_interval(nums, intervals):
    seq = [classify_interval(x, intervals) for x in nums]
    m = len(intervals)
    L = len(seq)
    if L == 0:
        return [0] * m

    run_values = []
    for target in range(m):
        if seq[-1] == target:
            r = 0
            j = L - 1
            while j >= 0 and seq[j] == target:
                r += 1
                j -= 1
            run_values.append(+r)
        else:
            g = 0
            j = L - 1
            while j >= 0 and seq[j] != target:
                g += 1
                j -= 1
            run_values.append(-g)
    return run_values

# ----- 시뮬레이션 + 상세 출력 -----
def simulate_run_value_counts(n_jackpots): # 10000은 형식적인 자리표시자란다(n_jackpots=10000에서 10000지워버림)
    jackpot_nums = []
    value_counter = Counter()

    for idx in range(1, n_jackpots + 1):
        jackpot = make_jackpot_num()
        jackpot_nums.append(jackpot)

        seq = [classify_interval(x, intervals) for x in jackpot_nums]

        print(f"\n🔥 {idx}번째 잭팟 발생 → 값 = {jackpot}")
        print(f"구간 시퀀스(seq): {seq}")

        if len(jackpot_nums) >= 5:
            run_values = signed_run_length_by_interval(jackpot_nums, intervals)
            print(f"연속도(run values): {run_values}\n")

            for val in run_values:
                value_counter[val] += 1

    return value_counter


# ----- 실행 -----
n = int(input("생성할 잭팟 개수를 입력하세요: "))
value_counts = simulate_run_value_counts(n)

# ----- DataFrame 변환 -----
total = sum(value_counts.values())
df = pd.DataFrame(
    [(val, count, count / total * 100) for val, count in value_counts.items()],
    columns=["Run Value(연속도)", "Count", "Probability(%)"]
).sort_values("Probability(%)", ascending=False).reset_index(drop=True)

# ----- 출력 -----
pd.set_option("display.max_rows", None)
print("\n=== 연속도 값별 등장 확률 (내림차순) ===")
print(df)
print(f"\n✅ 총 {len(df)}가지 연속도 값이 등장했습니다. (전체 {total}회)")
```

## 구간연속지표에서 각 구간이 모두 발생확률이 같은 균등분포임을 이용해 (1/5)^n과같이 확률을 구하는법 적용

```python
import random
import math

# ----- 확률 파라미터 -----
p = 1 / 125
q = 1 - p

total = int(input("얼마를 충전하시겠습니까? "))
jackpot_num_list = []
previous_betting = 0

# 구간 정의(양끝 포함)
intervals = [(1, 28), (29, 64), (65, 115), (116, 200), (201, float('inf'))]


def make_jackpot_num():
    """기하분포(p=1/125)를 시뮬레이션해서 첫 잭팟 시점을 반환"""
    jackpot_num = 1
    while True:
        if random.randint(1, 125) == 1:
            return jackpot_num
        jackpot_num += 1


def classify_interval(x, intervals):
    """값 x가 속한 구간의 인덱스(0~len-1)를 반환"""
    for i, (lo, hi) in enumerate(intervals):
        if lo <= x <= hi:
            return i
    return None


def signed_streaks_by_interval(nums, intervals):
    """각 구간의 '부호 있는 스트릭' 계산"""
    seq = [classify_interval(x, intervals) for x in nums]
    m = len(intervals)
    L = len(seq)
    if L == 0:
        return [0] * m

    streaks = []
    for target in range(m):
        if seq[-1] == target:
            r = 0
            j = L - 1
            while j >= 0 and seq[j] == target:
                r += 1
                j -= 1
            streaks.append(+r)
        else:
            g = 0
            j = L - 1
            while j >= 0 and seq[j] != target:
                g += 1
                j -= 1
            streaks.append(-g)
    return streaks


def streak_to_probability(streak):
    """구간 스트릭을 확률로 변환"""
    p = 1 / 5
    q = 4 / 5

    if streak > 0:      # 양수 → 연속 출현
        return p ** streak
    elif streak < 0:    # 음수 → 연속 미출현
        return q ** (-streak)
    else:
        return 1


def count_all_intervals(nums, intervals):
    """전체 잭팟 기록에서 각 구간의 누적 출현 횟수 계산"""
    counts = [0] * len(intervals)
    for x in nums:
        idx = classify_interval(x, intervals)
        if idx is not None:
            counts[idx] += 1
    return counts, len(nums)


def get_disappeared_bin(nums, intervals, window_size=5):
    """최근 window_size개 구간 중 사라진 구간 계산"""
    if len(nums) < window_size:
        return None
    old_value = nums[-window_size]
    disappeared_bin = classify_interval(old_value, intervals)
    return disappeared_bin


def moving_average(data, n):
    """최근 n개 평균"""
    if len(data) == 0:
        return 0
    subset = data[-n:] if len(data) >= n else data
    return round(sum(subset) / len(subset))


def interval_distribution(nums, intervals, n):
    """최근 n개 구간 분포"""
    recent = nums[-n:]
    counts = [0] * len(intervals)
    for num in recent:
        idx = classify_interval(num, intervals)
        if idx is not None:
            counts[idx] += 1
    return counts


def slot_machine():
    global total, previous_betting, jackpot_num_list
    jackpot_num = make_jackpot_num()
    play_num = 0

    while True:
        betting_str = input("얼마를 배팅 하시겠습니까? ")

        if betting_str == '':
            if previous_betting == 0:
                print("이전 배팅 금액이 없습니다. 배팅 금액을 입력해주세요.")
                continue
            else:
                betting = previous_betting
        else:
            try:
                betting = int(betting_str)
                if betting <= 0:
                    print("배팅 금액은 0보다 커야 합니다.")
                    continue
            except ValueError:
                print("유효한 배팅 금액을 입력해주세요.")
                continue

        if betting > total:
            print("배팅 금액이 잔액을 초과합니다.")
            continue

        previous_betting = betting
        play_num += 1
        total -= betting

        probability = (1 - (q ** play_num)) * 100

        # 💡 구간분포: 5, 10, 15, 20개 기준으로 계산
        interval_counts = {n: interval_distribution(jackpot_num_list, intervals, n) for n in [5, 10, 15, 20]}

        # 💡 이동평균: 5, 10, 15, 20개 기준으로 계산
        averages = {n: moving_average(jackpot_num_list, n) for n in [5, 10, 15, 20]}

        signed_streaks = signed_streaks_by_interval(jackpot_num_list, intervals)
        total_counts, n_total = count_all_intervals(jackpot_num_list, intervals)

        disappeared_bin = get_disappeared_bin(jackpot_num_list, intervals, window_size=5)

        print("\n")
        print(f"현재 {play_num}회 배팅 중입니다. total: {total}")
        for n in [5, 10, 15, 20]:
            print(f"구간분포(최근{n}) {interval_counts[n]}")
        print(f"당첨확률(기하CDF): {probability:.2f}%")

        avg_str = ", ".join([f"{n}개={averages[n]}" for n in [5, 10, 15, 20]])
        print(f"최근 평균: {avg_str}")

        # 🔥 추가된 기능: 구간별 연속 확률 출력
        print(f"구간별 연속정도 : {signed_streaks}")

        streak_probs = [streak_to_probability(s) for s in signed_streaks]
        streak_probs_percent = [round(p * 100, 6) for p in streak_probs]

        print(f"구간별 연속확률 : {streak_probs_percent}%")

        print(f"전체 {n_total}개 구간별 누적 횟수: {total_counts}")

        if disappeared_bin is not None:
            lo, hi = intervals[disappeared_bin]
            print(f"최근 5회 기준 사라진 구간: {disappeared_bin + 1} ({lo}~{hi})")
        else:
            print(f"최근 5회 기준 사라진 구간: 없음 (데이터 부족)")

        print(f"(1,28)(29,64)(65,115)(116,200)(201), 이전 {jackpot_num_list[-5:]}")

        # 🎰 잭팟 처리
        if play_num == jackpot_num:
            total += betting * 110
            jackpot_num_list.append(play_num)
            print(f"이야~ 대박!🧞‍♂️ 축하합니다! 잭팟입니다! total: {total}")
            print("🧚‍♂️ 🧚‍♀️")
            break


while True:
    slot_machine()
```

## 소프트맥스를 이용해 해석이 어려운 확률을 다음판에 그곳에서 당첨될 확률이 어느정도인지를 알려주는 지표로 바꾸기
- 이때 핵심은 소프트맥스를 적용하면서 기존확률계산 공식이 아니라 likelihood개념과 같이 날것의 강도를 표현해주는 지표로 만들어 인공지능 신경망에 적용하거나 인공지능 신경망이 어떻게 확률을 이해하고 계산하는지 더잘 이해하고 적용해보는 연습이 될수있는 상황으로 바뀌었다

```python
import random
import math

# ----- 확률 파라미터 -----
p = 1 / 125
q = 1 - p

total = int(input("얼마를 충전하시겠습니까? "))
jackpot_num_list = []
previous_betting = 0

# 구간 정의(양끝 포함)
intervals = [(1, 28), (29, 64), (65, 115), (116, 200), (201, float('inf'))]


def make_jackpot_num():
    """기하분포(p=1/125)를 시뮬레이션해서 첫 잭팟 시점을 반환"""
    jackpot_num = 1
    while True:
        if random.randint(1, 125) == 1:
            return jackpot_num
        jackpot_num += 1


def classify_interval(x, intervals):
    """값 x가 속한 구간의 인덱스(0~len-1)를 반환"""
    for i, (lo, hi) in enumerate(intervals):
        if lo <= x <= hi:
            return i
    return None


def signed_streaks_by_interval(nums, intervals):
    """각 구간의 '부호 있는 스트릭' 계산"""
    seq = [classify_interval(x, intervals) for x in nums]
    m = len(intervals)
    L = len(seq)
    if L == 0:
        return [0] * m

    streaks = []
    for target in range(m):
        if seq[-1] == target:
            r = 0
            j = L - 1
            while j >= 0 and seq[j] == target:
                r += 1
                j -= 1
            streaks.append(+r)
        else:
            g = 0
            j = L - 1
            while j >= 0 and seq[j] != target:
                g += 1
                j -= 1
            streaks.append(-g)
    return streaks


def streak_to_probability(streak):
    """run-length로 likelihood 계산"""
    p = 1 / 5
    q = 4 / 5

    if streak > 0:      # 양수 → 연속 출현 likelihood
        return p ** streak
    elif streak < 0:    # 음수 → 연속 미출현 likelihood
        return q ** (-streak)
    else:
        return 1


def softmax(weights):
    """weight 리스트를 softmax 확률로 변환"""
    total = sum(weights)
    if total == 0:
        return [1 / len(weights)] * len(weights)
    return [w / total for w in weights]


def count_all_intervals(nums, intervals):
    counts = [0] * len(intervals)
    for x in nums:
        idx = classify_interval(x, intervals)
        if idx is not None:
            counts[idx] += 1
    return counts, len(nums)


def get_disappeared_bin(nums, intervals, window_size=5):
    if len(nums) < window_size:
        return None
    old_value = nums[-window_size]
    disappeared_bin = classify_interval(old_value, intervals)
    return disappeared_bin


def moving_average(data, n):
    if len(data) == 0:
        return 0
    subset = data[-n:] if len(data) >= n else data
    return round(sum(subset) / len(subset))


def interval_distribution(nums, intervals, n):
    recent = nums[-n:]
    counts = [0] * len(intervals)
    for num in recent:
        idx = classify_interval(num, intervals)
        if idx is not None:
            counts[idx] += 1
    return counts


def slot_machine():
    global total, previous_betting, jackpot_num_list
    jackpot_num = make_jackpot_num()
    play_num = 0

    while True:
        betting_str = input("얼마를 배팅 하시겠습니까? ")

        if betting_str == '':
            if previous_betting == 0:
                print("이전 배팅 금액이 없습니다. 배팅 금액을 입력해주세요.")
                continue
            else:
                betting = previous_betting
        else:
            try:
                betting = int(betting_str)
                if betting <= 0:
                    print("배팅 금액은 0보다 커야 합니다.")
                    continue
            except ValueError:
                print("유효한 배팅 금액을 입력해주세요.")
                continue

        if betting > total:
            print("배팅 금액이 잔액을 초과합니다.")
            continue

        previous_betting = betting
        play_num += 1
        total -= betting

        probability = (1 - (q ** play_num)) * 100

        interval_counts = {n: interval_distribution(jackpot_num_list, intervals, n) for n in [5, 10, 15, 20]}
        averages = {n: moving_average(jackpot_num_list, n) for n in [5, 10, 15, 20]}

        signed_streaks = signed_streaks_by_interval(jackpot_num_list, intervals)
        streak_likelihoods = [streak_to_probability(s) for s in signed_streaks]

        # ⭐ 핵심 변경:
        # likelihood → 역수 weight → softmax
        weights = [1 / L for L in streak_likelihoods]
        predicted_probs = softmax(weights)

        total_counts, n_total = count_all_intervals(jackpot_num_list, intervals)
        disappeared_bin = get_disappeared_bin(jackpot_num_list, intervals, window_size=5)

        print("\n")
        print(f"현재 {play_num}회 배팅 중입니다. total: {total}")
        for n in [5, 10, 15, 20]:
            print(f"구간분포(최근{n}) {interval_counts[n]}")
        print(f"당첨확률(기하CDF): {probability:.2f}%")

        avg_str = ", ".join([f"{n}개={averages[n]}" for n in [5, 10, 15, 20]])
        print(f"최근 평균: {avg_str}")

        print(f"구간별 연속정도 : {signed_streaks}")
        print(f"구간별 likelihood : {[round(l, 6) for l in streak_likelihoods]}")
        print(f"역수 weight : {[round(w, 6) for w in weights]}")

        print(f"다음 판 구간별 예측 확률(softmax, 역수 기반) : {[round(p * 100, 2) for p in predicted_probs]}%")

        print(f"전체 {n_total}개 구간별 누적 횟수: {total_counts}")

        if disappeared_bin is not None:
            lo, hi = intervals[disappeared_bin]
            print(f"최근 5회 기준 사라진 구간: {disappeared_bin + 1} ({lo}~{hi})")
        else:
            print("최근 5회 기준 사라진 구간: 없음 (데이터 부족)")

        print(f"(1,28)(29,64)(65,115)(116,200)(201), 이전 {jackpot_num_list[-5:]}")

        if play_num == jackpot_num:
            total += betting * 110
            jackpot_num_list.append(play_num)
            print(f"이야~ 대박!🧞‍♂️ 축하합니다! 잭팟입니다! total: {total}")
            print("🧚‍♂️ 🧚‍♀️")
            break


while True:
    slot_machine()
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt

# ====== 1. 데이터 ======
X = torch.tensor([[0.,0.],
                  [0.,1.],
                  [1.,0.],
                  [1.,1.]], dtype=torch.float32)
y = torch.tensor([[0.],
                  [1.],
                  [1.],
                  [0.]], dtype=torch.float32)


# ====== 2. 모델 ======
class XORNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(2, 2)
        self.act1 = nn.Tanh()
        self.fc2 = nn.Linear(2, 1)
        self.act2 = nn.Sigmoid()

    def forward(self, x):
        h = self.act1(self.fc1(x))
        out = self.act2(self.fc2(h))
        return out, h

model = XORNet()


# ====== 3. 손실 & 옵티마이저 ======
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.1)

# ====== 4. 기록 ======
epochs = 5000
log_loss = []
log_W1 = []
log_W2 = []
log_yhat = []
log_hidden = []

# ====== 5. 학습 ======
for epoch in range(epochs):

    y_hat, hidden = model(X)
    loss = criterion(y_hat, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    # 기록 (매 epoch 저장)
    log_loss.append(loss.item())
    log_W1.append(model.fc1.weight.data.clone().numpy())
    log_W2.append(model.fc2.weight.data.clone().numpy())
    log_yhat.append(y_hat.data.clone().numpy())
    log_hidden.append(hidden.data.clone().numpy())


# ====== 6. 시각화 ======
plt.figure(figsize=(15, 10))

# ---- (1) Loss 그래프 ----
plt.subplot(2, 2, 1)
plt.plot(log_loss)
plt.title("Loss Curve")
plt.xlabel("Epoch")
plt.ylabel("Loss")

# ---- (2) W1 변화 ----
W1_arr = torch.tensor(log_W1).numpy()  # (epochs, 2, 2)
plt.subplot(2, 2, 2)
plt.plot(W1_arr[:,0,0], label="W1[0,0]")
plt.plot(W1_arr[:,0,1], label="W1[0,1]")
plt.plot(W1_arr[:,1,0], label="W1[1,0]")
plt.plot(W1_arr[:,1,1], label="W1[1,1]")
plt.title("Hidden Layer Weights (W1)")
plt.xlabel("Epoch")
plt.legend()

# ---- (3) W2 변화 ----
W2_arr = torch.tensor(log_W2).numpy()  # (epochs, 1, 2)
plt.subplot(2, 2, 3)
plt.plot(W2_arr[:,0,0], label="W2[0]")
plt.plot(W2_arr[:,0,1], label="W2[1]")
plt.title("Output Layer Weights (W2)")
plt.xlabel("Epoch")
plt.legend()

# ---- (4) 출력값 변화 ----
yhat_arr = torch.tensor(log_yhat).numpy()  # (epochs, 4, 1)
plt.subplot(2, 2, 4)
plt.plot(yhat_arr[:,0,0], label="y_hat(0,0)")
plt.plot(yhat_arr[:,1,0], label="y_hat(0,1)")
plt.plot(yhat_arr[:,2,0], label="y_hat(1,0)")
plt.plot(yhat_arr[:,3,0], label="y_hat(1,1)")
plt.title("Prediction Change per Input")
plt.xlabel("Epoch")
plt.legend()

plt.tight_layout()
plt.show()


# ====== 7. 최종결과 ======
final_y, final_h = model(X)
print("Final Output:\n", final_y)
print("Final Hidden:\n", final_h)
print("Final Loss:", criterion(final_y, y).item())

'''
풀배치수로 에포크수가 네개의 입력값이 모두들어가서 몇번학습됐는지 수와 같다.
'''
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt

torch.manual_seed(2)  # 랜덤 시드 고정

# ====== 1. 데이터 ======
X = torch.tensor([[0.,0.],
                  [0.,1.],
                  [1.,0.],
                  [1.,1.]], dtype=torch.float32)
y = torch.tensor([[0.],
                  [1.],
                  [1.],
                  [0.]], dtype=torch.float32)


# ====== 2. 모델 ======
class XORNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(2, 2)
        self.act1 = nn.Tanh()
        self.fc2 = nn.Linear(2, 1)
        self.act2 = nn.Sigmoid()

        # ---- 초기 가중치 강제 설정 ---- # 위에서 초기가중치 랜덤시드를 주어도 아래값들로 초기가중치 고정
        with torch.no_grad():
            self.fc1.weight.copy_(torch.tensor([[0.5, -0.3],
                                                [-0.7, 0.4]]))
            self.fc1.bias.copy_(torch.tensor([0.0, 0.0]))

            self.fc2.weight.copy_(torch.tensor([[0.2, -0.1]]))
            self.fc2.bias.copy_(torch.tensor([0.0]))

    def forward(self, x):
        h = self.act1(self.fc1(x))
        out = self.act2(self.fc2(h))
        return out, h

model = XORNet()


# ====== 3. 손실 & 옵티마이저 ======
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.1)

# ====== 4. 기록 ======
epochs = 5000
log_loss = []
log_W1 = []
log_W2 = []
log_yhat = []
log_hidden = []


# ====== 5. 학습 ======
for epoch in range(epochs):

    y_hat, hidden = model(X)
    loss = criterion(y_hat, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    log_loss.append(loss.item())
    log_W1.append(model.fc1.weight.data.clone().numpy())
    log_W2.append(model.fc2.weight.data.clone().numpy())
    log_yhat.append(y_hat.data.clone().numpy())
    log_hidden.append(hidden.data.clone().numpy())


# ====== 6. 그래프 출력 ======
plt.figure(figsize=(15, 10))

plt.subplot(2,2,1)
plt.plot(log_loss); plt.grid(True); plt.title("Loss")

plt.subplot(2,2,2)
W1_arr = torch.tensor(log_W1).numpy()
plt.plot(W1_arr[:,0,0], label="W1[0,0]")
plt.plot(W1_arr[:,0,1], label="W1[0,1]")
plt.plot(W1_arr[:,1,0], label="W1[1,0]")
plt.plot(W1_arr[:,1,1], label="W1[1,1]")
plt.grid(True); plt.title("W1"); plt.legend()

plt.subplot(2,2,3)
W2_arr = torch.tensor(log_W2).numpy()
plt.plot(W2_arr[:,0,0], label="W2[0]")
plt.plot(W2_arr[:,0,1], label="W2[1]")
plt.grid(True); plt.title("W2"); plt.legend()

plt.subplot(2,2,4)
yh = torch.tensor(log_yhat).numpy()
plt.plot(yh[:,0,0], label="(0,0)")
plt.plot(yh[:,1,0], label="(0,1)")
plt.plot(yh[:,2,0], label="(1,0)")
plt.plot(yh[:,3,0], label="(1,1)")
plt.grid(True); plt.title("y_hat"); plt.legend()

plt.tight_layout()
plt.show()
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt

torch.manual_seed(42)

# ====== 1. 데이터 ======
X = torch.tensor([[0.,0.],
                  [0.,1.],
                  [1.,0.],
                  [1.,1.]], dtype=torch.float32)
y = torch.tensor([[0.],
                  [1.],
                  [1.],
                  [0.]], dtype=torch.float32)

# ====== 2. 모델 ======
class XORNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(2, 2)
        self.act1 = nn.ReLU()
        self.fc2 = nn.Linear(2, 1)
        self.act2 = nn.ReLU()

        with torch.no_grad():
            self.fc1.weight.copy_(torch.tensor([[0.5, -0.3],
                                                [-0.7, 0.4]]))
            self.fc1.bias.copy_(torch.tensor([0.0, 0.0]))

            self.fc2.weight.copy_(torch.tensor([[0.2, -0.1]]))
            self.fc2.bias.copy_(torch.tensor([0.0]))

    def forward(self, x):
        h = self.act1(self.fc1(x))
        out = self.act2(self.fc2(h))
        return out, h

model = XORNet()

# ====== 3. 손실 & 옵티마이저 ======
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.1)

# ====== 4. 기록 ======
epochs = 200
log_loss = []
log_W1 = []
log_W2 = []
log_b1 = []
log_b2 = []
log_yhat = []
log_hidden = []

# ====== 5. 학습 ======
for epoch in range(epochs):
    y_hat, hidden = model(X)
    loss = criterion(y_hat, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    log_loss.append(loss.item())
    log_W1.append(model.fc1.weight.data.clone().numpy())
    log_W2.append(model.fc2.weight.data.clone().numpy())
    log_b1.append(model.fc1.bias.data.clone().numpy())
    log_b2.append(model.fc2.bias.data.clone().numpy())
    log_yhat.append(y_hat.data.clone().numpy())
    log_hidden.append(hidden.data.clone().numpy())


# ====== 6. 그래프 출력 ======
plt.figure(figsize=(15, 15))

# ---- 1. Loss ----
plt.subplot(3, 2, 1)
plt.plot(log_loss); plt.grid(True)
plt.title("Loss")

# ---- 2. W1 ----
plt.subplot(3, 2, 2)
W1_arr = torch.tensor(log_W1).numpy()
plt.plot(W1_arr[:,0,0], label="W1[0,0]")
plt.plot(W1_arr[:,0,1], label="W1[0,1]")
plt.plot(W1_arr[:,1,0], label="W1[1,0]")
plt.plot(W1_arr[:,1,1], label="W1[1,1]")
plt.legend(); plt.grid(True); plt.title("W1")

# ---- 3. W2 ----
plt.subplot(3, 2, 3)
W2_arr = torch.tensor(log_W2).numpy()
plt.plot(W2_arr[:,0,0], label="W2[0]")
plt.plot(W2_arr[:,0,1], label="W2[1]")
plt.legend(); plt.grid(True); plt.title("W2")

# ---- 4. y_hat ----
plt.subplot(3, 2, 4)
yh = torch.tensor(log_yhat).numpy()
plt.plot(yh[:,0,0], label="(0,0)")
plt.plot(yh[:,1,0], label="(0,1)")
plt.plot(yh[:,2,0], label="(1,0)")
plt.plot(yh[:,3,0], label="(1,1)")
plt.legend(); plt.grid(True); plt.title("y_hat")

# ---- 5. b1 ----
plt.subplot(3, 2, 5)
b1_arr = torch.tensor(log_b1).numpy()
plt.plot(b1_arr[:,0], label="b1[0]")
plt.plot(b1_arr[:,1], label="b1[1]")
plt.legend(); plt.grid(True); plt.title("Bias b1 (Hidden Layer)")

# ---- 6. b2 ----
plt.subplot(3, 2, 6)
b2_arr = torch.tensor(log_b2).numpy()
plt.plot(b2_arr[:,0], label="b2[0]")
plt.legend(); plt.grid(True); plt.title("Bias b2 (Output Layer)")

plt.tight_layout()
plt.show()

# ====== 7. 최종 가중치 출력 ======
print("\nFinal Weights and Biases")
print("W1:\n", model.fc1.weight.data)
print("b1:\n", model.fc1.bias.data)
print("\nW2:\n", model.fc2.weight.data)
print("b2:\n", model.fc2.bias.data)

# ====== 8. 최종 출력값 ======
final_y, final_h = model(X)
print("\nFinal Output (y_hat):\n", final_y)
print("\nFinal Hidden Output:\n", final_h)
print("\nFinal Loss:", criterion(final_y, y).item())
```
