```python
# 아무 지표 없는 단순한 게임 그자체 (슬롯머신 단순화 버전)
```

```python
import random

total = int(input("얼마를 충전하시겠습니까? "))

# 잭팟 번호 리스트를 전역 변수로 관리합니다.
jackpot_num_list = []

# 이전 배팅 금액을 저장하는 변수를 설정합니다. 초기 배팅 금액을 정의합니다.
previous_betting = 0

def slot_machine():
    global total, previous_betting  # 전역 변수 total와 previous_betting를 사용하겠다고 선언합니다.
    jackpot_num = random.randint(1, 125)
    play_num = 0

    while True:
        betting_str = input("얼마를 배팅 하시겠습니까?")
        if betting_str == '':
            if previous_betting == 0:
                print("이전 배팅 금액이 없습니다. 배팅 금액을 입력해주세요.")
                continue
            else:
                betting = previous_betting
        else:
            try:
                betting = int(betting_str)
                if betting <= 0:
                    print("배팅 금액은 0보다 커야 합니다.")
                    continue
            except ValueError:
                print("유효한 배팅 금액을 입력해주세요.")
                continue

        if betting > total:
            print("배팅 금액이 잔액을 초과합니다.")
            continue

        previous_betting = betting  # 현재 배팅 금액을 기억합니다.
        play_num += 1
        total -= betting
        print("\n")
        print(f"현재 {play_num}회 배팅 중입니다. total: {total}")

        if play_num == jackpot_num:
            total += betting * 110
            print(f"축하합니다! 잭팟입니다! total: {total}")
            jackpot_num_list.append(play_num)
            break

while True:
    print(f"이전 잭팟까지의 횟수 = {jackpot_num_list}")
    slot_machine()
```
